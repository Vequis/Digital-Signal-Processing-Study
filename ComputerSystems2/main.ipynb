Testing Complex Operations (Check C libraries)



```python
import numpy as np

# Gerar um número complexo aleatório
z = 0.5 + 1j * 0.8

# Definir valores para k e teta
teta = np.pi / 3

# Calcular e^(-k * teta)
factor = np.exp(1j * teta)

# Multiplicar o número complexo pelo fator
result = z * factor

print("Número complexo:", z)
print("Resultado da multiplicação:", result)
```
