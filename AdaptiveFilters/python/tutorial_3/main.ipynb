```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def NLMS(x, h, N, L, mi):
    # What is L?
    h_hat = np.zeros(N)

    y = np.convolve(x, h)

    x_padded = np.concatenate((np.zeros(N-1), x))

    if L is None or L > len(x):
        L = len(x)

    errors = np.zeros(L)
    mismatches = np.zeros(L)
    for i in range(L):
        # Should I start with x_v = [x(N-1) x(N-2) ... x(0)] ?
        x_v = x_padded[i:i+N][::-1]
        # print()
        print(f"Iteration {i + 1}")

        # print(f"y: {y}")
        # print(f"x_v: {x_v}")

        y_hat = np.dot(h_hat, x_v)
        e = y[i] - y_hat
        # print(f"y_hat: {y_hat}")
        # print(f"h_hat: {h_hat}")
        # print(f"e: {e}")
        
        h_hat = h_hat + mi * e * x_v / (np.dot(x_v, x_v) + 1e-8)

        mismatch = np.linalg.norm(h[:N] - h_hat, ord=2)**2
        mismatches[i] = mismatch
        errors[i] = e

    return errors, mismatches, h_hat

x = np.random.randn(1000)
h = np.array([0.1, 0.2, 0.3, 0.2, 0.1])
y = np.convolve(x, h)

# print(x[0:5])

NLMS(x, h, 5, 200, 1)
```

```python
def ar1(x, corr=0.9):
    """
    Correlate an input by using an autoregressive (AR) process of order 1.

    Parameters
    ----------
    x : array_like
        Input signal.
    corr : float, optional
        Correlation coefficient (default is 0.9).

    Returns
    -------
    x_ar : np.ndarray
        Correlated output signal.
    """
    x_ar = np.zeros_like(x)
    x_ar[0] = x[0]
    for i in range(1, len(x)):
        x_ar[i] = x[i] + corr * x_ar[i - 1]
    return x_ar

x_white = white_signal(6000)
x_colored = ar1(x)

# Compute Fourier Transform
X_white = np.fft.fft(x_white)
X_colored = np.fft.fft(x_colored)

# Frequency axis
freqs = np.fft.fftfreq(len(x_white), d=1)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(freqs[:len(freqs)//2], np.abs(X_white)[:len(freqs)//2])
plt.title('Fourier Transform of White Signal')
plt.xlabel('Frequency')
plt.ylabel('Magnitude')

plt.subplot(1, 2, 2)
plt.plot(freqs[:len(freqs)//2], np.abs(X_colored)[:len(freqs)//2])
plt.title('Fourier Transform of Colored Signal')
plt.xlabel('Frequency')
plt.ylabel('Magnitude')

plt.tight_layout()
plt.show()

```
